## DBSCAN

### Advantages of DBSCAN:

Is great at separating clusters of high density versus clusters of low density within a given dataset.

Is great with handling outliers within the dataset.

### Disadvantages of DBSCAN:

Does not work well when dealing with clusters of varying densities. While DBSCAN is great at separating high density clusters from low density clusters, DBSCAN struggles with clusters of similar density.

Struggles with high dimensionality data. I know, this entire article I have stated how DBSCAN can handle different dimensions and shapes. However, DBSCAN can only go so far, if given data with too many dimensions, DBSCAN suffers.

```python
# DBSCAN Clustering
# Importing the libraries

import numpy as np
import pandas as pd
```

```python
# Importing the dataset
dataset = pd.read_csv('Mall_Customers.csv')
X = dataset.iloc[:, [3, 4]].values
```

```python
dataset.head()
```

```python
# Using the elbow method to find the optimal number of clusters

from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps = 3, min_samples = 4)
```

```python
# Fitting the model

model = dbscan.fit(X)
```

```python
labels = model.labels_
```

```python
from sklearn import metrics

#identifying the points which makes up our core points
sample_cores = np.zeros_like(labels, dtype = bool)
```

```python
sample_cores[dbscan.core_sample_indices_] = True
```

```python
#Calculating the number of clusters

n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
```

```python
print(metrics.silhouette_score(X, labels))
```
